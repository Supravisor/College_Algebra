```python
# y = 1/8x + 31/16
!pip install matplotlib
!pip install numpy
import matplotlib.pyplot as plt
import numpy as np

xmin = -10
xmax = 10
ymin = -10
ymax = 10
points = 2*(xmax-xmin)
x = np.linspace(xmin, xmax, points)

fig, ax = plt.subplots()
plt.axis([xmin,xmax,ymin,ymax]) # window size
plt.plot([xmin,xmax],[0,0],'b') # blue x axis
plt.plot([0,0],[ymin,ymax], 'b') # blue y axis

ax.set_xlabel("x values")
ax.set_ylabel("y values")
ax.set_title("y = 1/8x + 31/16")
ax.grid(True)

ax.set_xticks(np.arange(xmin, xmax, 1))
ax.set_yticks(np.arange(ymin, ymax, 1))

y = 1/8*x + 31/16
plt.plot(x,y, label='y=1/8x + 31/16')
plt.plot([-10],[0.6875], 'ro', label='x = -10')
plt.plot([10],[3.1875], 'go', label='x = 10')
plt.legend()
plt.show()
```

```python
# y = 75x - 22.5
!pip install matplotlib
!pip install numpy
import matplotlib.pyplot as plt
import numpy as np

xmin = -1
xmax = 1
ymin = -35
ymax = 5
points = 2*(xmax-xmin)
x = np.linspace(xmin, xmax, points)

fig, ax = plt.subplots()
plt.axis([xmin,xmax,ymin,ymax]) # window size
plt.plot([xmin,xmax],[0,0],'b') # blue x axis
plt.plot([0,0],[ymin,ymax], 'b') # blue y axis

ax.set_xlabel("x values")
ax.set_ylabel("y values")
ax.set_title("y = 75x - 22.5")
ax.grid(True)

ax.set_xticks(np.arange(xmin, xmax, 1))
ax.set_yticks(np.arange(ymin, ymax, 1))

y = 75*x - 22.5
plt.plot(x,y, label='y=75x - 22.5')
plt.plot([-0.1],[-30], 'ro', label='x = -0.1')
plt.plot([0.1],[-15], 'go', label='x = 0.1')
plt.legend()
plt.show()
```

```python
# y = ax + b
import matplotlib.pyplot as plt
import numpy as np

xmin = -4
xmax = 4
ymin = -4
ymax = 4
points = 2*(xmax-xmin)
x = np.linspace(xmin, xmax, points)

fig, ax = plt.subplots()
plt.axis([xmin,xmax,ymin,ymax]) # window size
plt.plot([xmin,xmax],[0,0],'b') # blue x axis
plt.plot([0,0],[ymin,ymax], 'b') # blue y axis

ax.set_xlabel("x values")
ax.set_ylabel("y values")
ax.set_title("y = ax + b")
ax.grid(True)

ax.set_xticks(np.arange(xmin, xmax, 1))
ax.set_yticks(np.arange(ymin, ymax, 1))

a = 2*x + 3
plt.plot(x,a, label='y=2x + 3')
b = 2*x + 4
plt.plot(x,b, label='y=2x + 4')
c = 2*x - 4
plt.plot(x,c, label='y=2x - 4')
d = 2*x - 5
plt.plot(x,d, label='y=2x - 5')
plt.legend()
plt.show()
```
