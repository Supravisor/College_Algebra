# Graphing Systems

## Use numpy library instead of a loop

```python
!pip install matplotlib
!pip install numpy
import matplotlib.pyplot as plt
import numpy as np

# Set grid dimensions
xmin = -10
xmax = 10
ymin = -10
ymax = 10

# Define how many points to plot 
points = 10*(xmax-xmin)

# Define the array of x values once
x = np.linspace(xmin,xmax,points)

# Draw coordinate plane
fig, ax = plt.subplots()
plt.axis([xmin,xmax,ymin,ymax]) # window size
plt.plot([xmin,xmax],[0,0],'b') # blue x axis
plt.plot([0,0],[ymin,ymax], 'b') # blue y axis

# Line 1
y1 = 3*x
plt.plot(x, y1)

# Line 2
y2 = x**3
plt.plot(x, y2)

# Show grid
ax.grid(True)
plt.show()
```

## Shading and graphing inequalities

```python
!pip install matplotlib
!pip install numpy
import matplotlib.pyplot as plt
import numpy as np

# Set grid dimensions
xmin = -10
xmax = 10
ymin = -10
ymax = 10

# Define how many points to plot
points = 2*(xmax-xmin)

# Define the array of x values once
x = np.linspace(xmin,xmax,points)

# Draw coordinate plane
fig, ax = plt.subplots()
plt.axis([xmin,xmax,ymin,ymax]) # window size
plt.plot([xmin,xmax],[0,0],'b') # blue x axis
plt.plot([0,0],[ymin,ymax], 'b') # blue y axis

# Line 1
y1 = x+6
plt.plot(x, y1,'-')
plt.fill_between(x, y1, ymax, facecolor='red')

# Line 2
y2 = x+3
plt.plot(x, y2,'-')
plt.fill_between(x, y2, y1, facecolor='yellow')

# Line 3
y3 = x-1
plt.plot(x, y3)
plt.fill_between(x, y3, y2, facecolor='green')

# Line 4
y4 = x-4
plt.plot(x, y4)
plt.fill_between(x, y4, y3, facecolor='blue')

# Show grid
plt.show()
```
