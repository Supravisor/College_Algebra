## Verbal
1) Terry is skiing down a steep hill. Terry's elevation, E(t), in feet after t seconds is given by E(t) = 3000 - 70t.
Write a complete sentence describing Terry's starting elevation and how it is changing over time.

Terry begins at 3000 feet up the hill and then goes down the slope at 70 feet per second.

2) Jessica is walking home from a friend's house. After 2 minutes she is 1.4 miles from home. Twelve minutes after leaving, she is 0.9 miles from home. What is her rate in miles per hour?

One and a half miles per hour.

3) A boat is 100 miles away from the marina, sailing directly toward it at 10 miles per hour. Write an equation for the distance of the boat from the marina after t hours.

D(t) = 100 - 10t

4) If the graphs of two linear functions are perpendicular, describe the relationship between the slopes and the y-intercepts.

If the graphs of two linear functions are perpendicular, their slopes are negative reciprocals (or opposite reciprocals) of each other, while their y-intercepts have no specific relationship and can be anything.

5) If a horizontal line has the equation f(x) = a and a vertical line has the equation x = a, what is the point of the intersection? Explain why what you found is the point of intersection.

Because the x-coordinate must be (a) to lie on the vertical line, and the y-coordinate must be (a) to lie on the horizontal line, the only coordinate in the entire Cartesian plane where both conditions are true at the same time is (a, a).
